 Single-Dimension Multi-Aggregation
 Groupby  gives you clean, single-level columns instantly.
 pivot_table - forces a complex, nested MultiIndex column

```python
import pandas as pd

# Sample employee dataset
data = {
    'Department': ['HR', 'IT', 'IT', 'HR', 'IT', 'HR'],
    'Salary': [50000, 70000, 80000, 55000, 90000, 60000]
}
df=pd.DataFrame(data)
df
```

```python
groupby_result=df.groupby('Department')['Salary'].agg(['mean','count'])
groupby_result
```

```python
pivot_result=df.pivot_table(index='Department',values='Salary',aggfunc=['mean','count'])
pivot_result
```

Question 5: Custom Grouping Logic via Functions

```python
#import pandas as pd

# 1. Create a dataset with a DatetimeIndex
dates = pd.date_range(start='2026-08-01', periods=5, freq='D') 
# Aug 1 (Fri), Aug 2 (Sat), Aug 3 (Sun), Aug 4 (Mon), Aug 5 (Tue)

df_dates = pd.DataFrame({'Sales': [100, 150, 200, 110, 130]}, index=dates)
df_dates
```

```python
# 2. Define our custom grouping function
def is_weekend(timestamp):
    # .weekday() returns 5 for Saturday and 6 for Sunday
    if timestamp.weekday() >= 5:
        return 'Weekend'
    return 'Weekday'
```

```python
grouped_function=df_dates.groupby(is_weekend)['Sales'].mean()
grouped_function
```

```python
import pandas as pd
df1 = pd.DataFrame({
    'Region': ['East', 'East', 'West', 'West'],
    'Product': ['A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 250]
})
df1
```

```python
#using pivot
pivot_summary=df1.pivot_table(index='Region',columns='Product',values='Sales',aggfunc='sum')
pivot_summary
```

```python
#using groupby + unstack
groupby_summary=df1.groupby(['Region','Product'])['Sales'].mean().unstack('Product')
groupby_summary
```

CATEGORICAL GHOST COLUMNS -observed=True/False

```python
import pandas as pd
df2 = pd.DataFrame({
    'Group': pd.Categorical(['A', 'A', 'B'], categories=['A', 'B', 'C']),
    'Value': [10, 20, 30]
})
df2
```

```python
df2.groupby('Group',observed=True).sum()
```

```python
import pandas as pd
df3 = pd.DataFrame(
    {'Sales': [100, 200, 300, 400]}, 
    index=['Apple', 'Banana', 'Cat', 'Dog']
)
df3
```

```python
# Group by the LENGTH of the index word
df3.groupby(lambda x: len(x)).sum()
```

```python
# Group using a mapping dictionary
category_map={
    'Apple':'Fruit',
    'Banana':'Fruit',
    'Cat':'Animal',
    'Dog':'Animal'
}
df3.groupby(category_map).sum()
```

AGG,FILTER, TRANSFORM AND APPLY

```python
import pandas as pd
df_sales = pd.DataFrame({
    'Department': ['Store_A', 'Store_A', 'Store_A', 'Store_B', 'Store_B'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Sales': [100, 200, 300, 400, 100]
})
df_sales
```

```python
#agg() (or aggregate()) — Summarize Data
df_sales.groupby('Department')['Sales'].agg(['sum','mean'])
```

```python
#2. transform() — Broadcast Summary Back to Original Rows
#Let's calculate each employee's contribution as a percentage of their department's total sales.
#Step 1: Calculate department total sales and broadcast it back
df_sales['Dep_Total']=df_sales.groupby('Department')['Sales'].transform('sum')
# Step 2: Calculate percentage
df_sales['Pectentage']=(df_sales['Sales']/df_sales['Dep_Total'])*100
df_sales
```

```python
#3. filter() — Keep or Drop Entire Groups
#Suppose we want to keep only the stores whose average sales are greater than 220.
filtered_df=df_sales.groupby('Department').filter(lambda x: x['Sales'].mean()>220)
filtered_df
```

OR OR OR 

```python
df_sales['average_sales']=df_sales.groupby('Department')['Sales'].transform('mean')
df_sales
```

```python
df_filtered=df_sales[df_sales['average_sales']>220]
df_filtered
```

4. apply() — The Flexible Swiss Army Knife

```python
#4. apply() — The Flexible Swiss Army Knife
def get_top_2(x):
    return x.sort_values(by='Sales',ascending=False).head(2)

#fetch department, top employee and sales 
top_performers=df_sales.groupby('Department',group_keys=False).apply(get_top_2,include_groups=True)

#fetch the Top employee and the sales 
#top_performers=df_sales.groupby('Department',group_keys=False)[['Employee','Sales']].apply(get_top_2)
top_performers
```

USING NLARGEST()

```python
top_performers=df_sales.groupby('Department',as_index=False).apply(
    lambda x: x.nlargest(3,'Sales'))

top_performers
```

NSMALLEST 

```python
top_performers=df_sales.groupby('Department',as_index=False).apply(
    lambda x: x.nsmallest(2,'Sales'))

top_performers
```
