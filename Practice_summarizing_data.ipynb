```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv(r"C:\Users\pchepkoech\OneDrive - CIHEB- KENYA\DRIVE D\DATA FY2025\July 2026\Lumumba_Active_on_ART.csv")
df.columns
```

```python
df['Last VL Result']=pd.to_numeric(df['Last VL Result'],errors='coerce')
```

```python
df['Last VL Result'].dtype
```

```python
def vl_cat(vl):
    if vl<50:
        return '<50 copies/ml'
    if vl<199:
        return '<200 copies/ml'
    if vl<999:
        return '<1000 copies<ml'
    else: 
        return 'viremic'

df['VL_Category']=df['Last VL Result'].apply(vl_cat)
df['VL_Category'].value_counts()
```

```python
bins = [0,10,15,20,120]
labels = ['0-9','10-14','15-19','20+']
df['VL_AgeBand']=pd.cut(df['Age at reporting'], bins=bins,labels=labels,right=False)
```

```python
#crosstab
pd.crosstab(index=df['VL_Category'],columns=df['VL_AgeBand'])
```

```python
#groupby

grouped=(df.groupby(['VL_AgeBand','VL_Category'],observed=False)[['Last VL Result']].count()
.unstack(fill_value=0)
)
grouped
```

```python
#pivot table 
df.pivot_table(index='VL_Category', columns='VL_AgeBand',values='Last VL Result',aggfunc=['count','mean'],observed=False)
```
